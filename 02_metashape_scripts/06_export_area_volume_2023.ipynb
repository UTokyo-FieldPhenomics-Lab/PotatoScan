```python
import Metashape
import pandas as pd
import numpy as np

import configs as cfg
import os
```

```python
ms_list = ['1R_Group0', '1R_Group1', '1R_Group2', 
           '2R_Group0',
           '3R_Group0', '3R_Group1',
           '4R_Group0', '4R_Group1',
           '5R_Group0', 
           'XR_GroupFix']
```

```python
vol_dict = {}
area_dict = {}

for ms_name in ms_list:
    doc = Metashape.Document()
    doc.open(os.path.join(cfg.working_directory, "projects.psx", ms_name + '.psx'))

    for chunk in doc.chunks:
        vol_dict[chunk.label] = chunk.model.volume()
        area_dict[chunk.label] = chunk.model.area()
```

read scalebarlist.csv to get the time-series order

```python
sbl = pd.read_csv('scalebarlist.csv', header=None)
sbl
```

```python
area_volume_csv = pd.DataFrame(columns=[['label', 'group_id', 'bucket_id', 'pin_id', 'area (cm2)', 'volume (cm3)']])

for i, label in enumerate(sbl[0]):
    vol_cm = vol_dict[label] * 1000 * 1000
    area_cm = area_dict[label] * 100 * 100

    if vol_cm == 0:
        print(f"{label} has 0 volume")

    if area_cm == 0:
        print(f"{label} has 0 area")

    area_volume_csv.loc[len(area_volume_csv),:] = [label, sbl.loc[i, 2], sbl.loc[i, 3], sbl.loc[i, 4], area_cm, vol_cm]
```
