```python
import Metashape
import pandas as pd
import numpy as np

import configs as cfg
import os

from pathlib import Path
```

```python
project_path = Path(cfg.working_directory) / "projects.psx"
```

```python
psx_list = [psx for psx in project_path.glob('**/*.psx') if psx.is_file()]
```

```python
vol_dict = {}
area_dict = {}

for psx in psx_list:
    doc = Metashape.Document()
    doc.open(str(psx))

    for chunk in doc.chunks:
        vol_dict[chunk.label] = chunk.model.volume()
        area_dict[chunk.label] = chunk.model.area()
```

```python
vol_dict['000'], type(vol_dict['001']), area_dict['002'], type(area_dict['003'])
```

```python
area_volume_csv = pd.DataFrame(columns=[['label', 'area (cm2)', 'volume (cm3)']])

for label in vol_dict.keys():
    vol_cm = vol_dict[label] * 1000 * 1000
    area_cm = area_dict[label] * 100 * 100

    if vol_cm == 0:
        print(f"{label} has 0 volume")

    if area_cm == 0:
        print(f"{label} has 0 area")

    area_volume_csv.loc[len(area_volume_csv),:] = [label, area_cm, vol_cm]

area_volume_csv
```

```python
area_volume_csv.to_csv( Path(cfg.working_directory) / "outputs" / "mesh_area_volume.csv", index=None)
```
